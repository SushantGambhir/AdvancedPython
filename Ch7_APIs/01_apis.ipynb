```python
import requests
import json
```

## GET

```python
url = 'https://jsonplaceholder.typicode.com/posts/1'

response = requests.get(url)

# Status code
print(response.status_code)
```

```python
# URL
print(response.url)
```

```python
# Header
print(response.headers)
```

```python
# Response body
print(response.text)
```

```python
print(response.json())
```

```python
data = response.json()
print(json.dumps(data, indent=2))
```
