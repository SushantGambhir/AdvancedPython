```python
from pydantic import BaseModel, Field, EmailStr
from typing import Optional, Literal

class employee_schema(BaseModel):
    name: str = Field(...,min_length=3,max_length=50) # ... represents that its mandatory and not optional
    age: int = Field(..., gt=18, lt=65) # Greater than, less than
    email: EmailStr = Field(..., description='Employee email address') # Will validate correct email format
    department: Optional[str] = None
    gender: Literal['Male','Female','Other'] = Field(...,description="Employee's gender") # Values can only be among these three values
```

```python
obj = employee_schema(name="John Doe", age=30, email="john.doe@example.com", department="IT", gender="Male")
obj
```

**Serialization in Pydantic is the process of converting Pydantic models into dictionaries or JSON strings, making them easier to store, transmit, or integrate with APIs.** It supports both **Python mode** (dict-like output) and **JSON mode** (JSON-compatible output), with fine-grained customization at the field or model level.  

```python
obj_dict=obj.model_dump()
obj_dict
```

```python
obj_json_str = obj.model_dump_json()
obj_json_str
```
