```python
from pydantic import BaseModel

# BaseModel is basically a class. We now need to create a class called schema where we shall define schema for our code.
# Then we inherit this BaseModel class which will then convert Schema class object into pydantic model.
```

```python
class employee_schema(BaseModel):
    name: str
    id: int
    department: str

# This is how we define class for Pydantic. No methods, no constructor, just attributes with data type
```

```python
obj = employee_schema(name='Sushant Gambhir', id=123, department='HR')
obj
```

```python
# We can also write the above line using dictionary, however we will have to unpack it
obj2 = employee_schema(**{"name": "Sushant Gambhir", "id": 123, "department": "HR"})
obj2
```

# **Coercion VS Strict Types**

### COERCION

### 🔑 What coercion does
- If you say a field is an `int`, and you pass `"123"` (string), Pydantic will **coerce** it into `123` (integer).  
- If you say a field is a `datetime`, and you pass `"2026-08-14 22:55"`, Pydantic will parse it into a proper `datetime` object.  
- If coercion isn’t possible (e.g., `"abc"` → `int`), Pydantic raises a `ValidationError`.

---


So in short: **coercion = automatic type conversion during validation**. It’s one of the main reasons Pydantic feels “smarter” than plain Python classes.  

```python
obj3 = employee_schema(**{"name": "John Doe", "id": "101", "department": "HR"})
obj3
```

### StrictTypes

No coercion here

```python
from pydantic import StrictInt

class employee_schema2(BaseModel):
    name: str
    id: StrictInt
    department: str
```

```python

# This will give error now

obj4 = employee_schema2(name = 'Sushant', id = '123', department= 'XYZ')
obj4
```

```python
# This will work

obj4 = employee_schema2(name = 'Sushant', id = 123, department= 'XYZ')
obj4
```
