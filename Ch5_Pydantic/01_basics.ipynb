```python
from pydantic import BaseModel

# BaseModel is basically a class. We now need to create a class called schema where we shall define schema for our code.
# Then we inherit this BaseModel class which will then convert Schema class object into pydantic model.
```

```python
class employee_schema(BaseModel):
    name: str
    id: int
    department: str

# This is how we define class for Pydantic. No methods, no constructor, just attributes with data type
```

```python
obj = employee_schema(name='Sushant Gambhir', id=123, department='HR')
obj
```
